```python
import nltk
from nltk.util import ngrams
```

```python
# 1. Generate Bigrams (pairs) and Trigrams (triplets)
bigrams_list = []
trigrams_list = []

for sublist in df_new['Conditions_Words']:
    bigrams_list.extend(list(ngrams(sublist, 2)))
    trigrams_list.extend(list(ngrams(sublist, 3)))

# 2. Count frequencies
bigram_counts = Counter(bigrams_list)
trigram_counts = Counter(trigrams_list)

print("--- Top 10 Most Common Bigrams ---")
for bg, count in bigram_counts.most_common(10):
    print(f"{' '.join(bg)}: {count}")

print("\n--- Top 10 Most Common Trigrams ---")
for tg, count in trigram_counts.most_common(10):
    print(f"{' '.join(tg)}: {count}")
```

```python
import sys
sys.path.append(".")

from collections import Counter

from pf1_dons.data_loader import load_catalog
from pf1_dons.models import OrGroup
from pf1_dons.engine import Character, filter_feats

catalog = load_catalog()
print("Dons chargés (hors #ERROR!):", len(catalog))

type_counts = Counter()
for feat in catalog:
    for req in feat.parsed.requirements:
        items = req.options if isinstance(req, OrGroup) else [req]
        for o in items:
            type_counts[o.type.value] += 1

for t, c in type_counts.most_common():
    print(f"{t}: {c}")

unparsed_feats = sum(1 for f in catalog if f.parsed.has_unparsed)
print(f"Dons avec au moins un segment UNPARSED: {unparsed_feats} ({100*unparsed_feats/len(catalog):.1f}%)")
```

```python
# Exemple : personnage de test -> liste de dons filtrée
perso = Character(
    character_class="Guerrier",
    level=5,
    race="nain",
    ability_scores={"Dex": 13, "For": 15, "Con": 14},
    known_feats={"Esquive"},
)
grouped = filter_feats(perso, catalog)
for status, results in grouped.items():
    print(status, len(results))

grouped["eligible"][:10]
```
